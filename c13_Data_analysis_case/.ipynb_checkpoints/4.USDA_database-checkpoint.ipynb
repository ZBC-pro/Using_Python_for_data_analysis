现在，为了对全部营养数据做分析，最简单的办法是将所有食物的营养成分整合到一张大表中。我们分成几个步骤来实现。

首先，将各食物的营养成分列表转换为DataFrame，并添加食物id的列；然后，将该DataFrame追加到一个列表中；最后，通过concat将这些数据拼接。

```python
# 用于存储列表中的营养元素
nutrients = []

# 遍历 db 中的每一个 json 记录
for rec in db:
    # 将每个json中的nutrients转化为DF单独提取出来
    fnuts = pd.DataFrame(rec["nutrients"])
    # 为 DF 赋值 id 列
    fnuts["id"] = rec["id"]
    # 将DF添加至nutrients列表
    nutrients.append(fnuts)

# 将 nutrients 列表中的所有 DF 纵向合并为一个 DataFrame
nutrients = pd.concat(nutrients, ignore_index=True)

nutrients.head()
```

我注意到这个DataFrame中存在一些重复项，直接丢弃就可以了

```python
# 计算 nutrients DataFrame 中重复行的数量
nutrients.duplicated().sum()
```

**两个DF分别是:info表，nutrients表。**  
info：description	group	id	manufacturer  
nutrient：value	units	description	group	id   

这两个表中都有description 和 group，所以需要区分

```python
# 去除重复项
nutrients = nutrients.drop_duplicates()

# 由于两个DF对象中都有 "group" 和 "description", 为了更加清晰，对其重命名：
col_mapping = {"description": "food",
               "group": "fgroup"}

info = info.rename(columns=col_mapping, copy=False)  # 在原数据上修改info列的重复列名

#查看是否修改成功
info.info()
```

```python
col_mapping = {"description": "nutrients",
               "group": "nutgroup"}

nutrients = nutrients.rename(columns=col_mapping, copy=False)  # 在原数据上修改列nutrients的重复列名

nutrients
```

```python
# 将info和nutrients合并：
ndata = pd.merge(nutrients, info, on="id")  # 通过id列内合并

# 查看合并后的列表
ndata.info()
```

```python
ndata.iloc[30000]  # 打印第30001条数据
```

我们现在可以根据食物分类和营养类型绘制一张中位值图

```python
ndata.head()
```

```python
# 将 ndata 按照营养成分（如锌、铁等）和食品组（如蔬菜、肉类等）分组，
# 对应的是value列中的值，并计算每个组中营养成分的中位数。
result = ndata.groupby(["nutrients", "fgroup"])["value"].quantile(0.5)

# 从结果中提取出所有食品组中锌（"Zinc, Zn") 含量的中位数。
result["Zinc, Zn"].sort_values().plot(kind="barh")

plt.title("根据食物分类得出锌的中位数")

plt.show()
```

```python
# 最后一列为中位数
result.head()
```

**使用Series的idmax或argmax方法，可以发现哪种食物最富含哪种营养。**

```python
# 对食物和营养数据进行分组，并找出每个营养分组中含有该营养成分最多的食物，
# 最后对食物名称进行截断，确保字符串长度不超过50个字符。

# nutgroup是富含的营养类型，nutrients是含有的营养成分
by_nutrient = ndata.groupby(["nutgroup", "nutrients"])

# 返回分组中最大的值
def get_maximum(x):
    return x.loc[x.value.idxmax()]

# 对value列和food列用函数找出最大值，即为最有营养的值
max_foods = by_nutrient.apply(get_maximum, include_groups=False)[["value", "food"]]

# 缩短食物字符串
max_foods = max_foods["food"].str[:50]

max_foods
```

by_nutrient = ndata.groupby(["nutgroup", "nutrients"])  
ndata 是一个包含食物和营养数据的 DataFrame，代码通过 groupby() 将数据按照 "nutgroup" 和 "nutrients" 进行分组。  
nutgroup：代表富含的营养类型（如蛋白质类、维生素类等）。  
nutrients：代表具体的营养成分（如锌、钙、铁等）。  
结果 by_nutrient 是一个按营养类型和营养成分分组的对象，后续会对每个分组进行操作。  



get_maximum()：  
get_maximum(x) 是一个用于从分组中找出含量最大的函数：  
x.value.idxmax() 返回 value 列中最大值的索引。   
x.loc[x.value.idxmax()]：基于这个最大值索引，返回该行的数据。  
这个函数的目的是找出每个分组中 "value" 列（营养成分的含量）的最大值所在的行，也就是含有该营养成分最多的食物。  



应用 get_maximum() 函数：  
max_foods = by_nutrient.apply(get_maximum, include_groups=False)[["value", "food"]]   
by_nutrient.apply(get_maximum, include_groups=False)：对每个分组应用 get_maximum() 函数，找出每个营养分组中 value 列最大的行，include_groups=False 用于忽略分组列在结果中的显示。  
[['value', 'food']]：选择结果中的 value 列（营养含量）和 food 列（食物名称），只保留这两列。  



max_foods = max_foods["food"].str[:50]  
max_foods["food"].str[:50]：从 food 列的字符串中，取前 50 个字符。这样做是为了防止食物名称过长影响显示。str[:50] 会将字符串截断到不超过 50 个字符。  



该代码首先按营养类型（nutgroup）和具体的营养成分（nutrients）对数据进行分组。  
然后在每个分组中，找出含有该营养成分最多的食物。  
最后将食物名称截断到 50 个字符以内，以方便显示。
